```python
%sql
SELECT DISTINCT _metadata.file_path, _metadata.file_size
FROM dev.bronze.sales
```

```python
%sql
SELECT
COUNT(*) as file_count,
ROUND(AVG(size_bytes) / 1e6, 2) as avg_file_size_mb
FROM (
SELECT _metadata.file_path, SUM(_metadata.file_size) as size_bytes
FROM dev.bronze.sales
GROUP BY _metadata.file_path
);
```
