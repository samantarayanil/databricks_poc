```python
df = (
    spark
    .read
    .format("csv")
    .option("header", "true")
    .load("/Volumes/dev/bronze/sales_vol/files/data.csv")
)

src_file = ['Australia', 'United Kingdom', 'Switzerland', 'Lithuania', 'Spain', 
           'Belgium', 'Japan', 'Poland', 'EIRE', 'Norway', 'Netherlands', 
           'Italy', 'France', 'Germany', 'Portugal', 'Denmark', 'Iceland', 
           'Channel Islands', 'Israel', 'Sweden', 'Austria', 'Cyprus', 
           'Finland', 'Bahrain', 'xyz', 'abc']
           
# Create src_file as DataFrame and use SQL except (distributed set difference)
src_df = spark.createDataFrame([(c,) for c in src_file], ["Country"])
extra_country_df = src_df.select("Country").exceptAll(df.select("Country").distinct())
extra_country = [row.Country for row in extra_country_df.collect()]  # Safe: only extra countries (small result)
extra_country
```
